```python
#fill the missing values of review_rating with median of every category 
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
#to know the sum of null values of all columns
df.isnull().sum()
```

```python
#to make readability of column
#first changing all the columns name to lowercase
df.columns=df.columns.str.lower()
```

```python
#replacing spaces with underscores
df.columns=df.columns.str.replace(" ","_")
```

```python
df.info()
```

```python
df=df.rename(columns={"purchase_amount_(usd)":"purchase_amount"})
```

```python
df.columns
```

```python
labels=['Young_Adult', 'Adult', 'Middle_aged','Senior']
```

```python
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age_group','age']]
```

```python
df['frequency_of_purchases']
```

```python
#purchase frequency day
frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Annually':365,
    'Bi-Weekly':14,
    'Monthly':30,
    'Quarterly':90,
    'Every 3 Months':90
}
df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```
